```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.svm import SVC
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
df=pd.read_csv("Placement.csv")
df.describe()
```

```python
df.drop(columns=["Student_ID"],inplace=True)
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
mm=MinMaxScaler()
x=mm.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lda=LinearDiscriminantAnalysis(n_components=1)
train_lda=lda.fit_transform(x_train,y_train)
test_lda=lda.transform(x_test)
```

```python
svc=SVC(kernel="poly")
svc.fit(train_lda,y_train)
```

```python
svc.score(test_lda,y_test)
```

```python
param_grid = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf', 'poly'],
    'gamma': [0.1, 1, 'scale', 'auto']
}
```

```python
gridcv=GridSearchCV(svc,param_grid,cv=5)
gridcv.fit(train_lda,y_train)
gridcv.score(test_lda,y_test)
```

```python
gridcv.best_params_
```

```python
ada=AdaBoostClassifier(SVC(C=10, gamma=1, kernel='rbf'),n_estimators=100,random_state=0)
ada.fit(train_lda,y_train)
```

```python
ada.score(test_lda,y_test)
```

```python
grad=GradientBoostingClassifier(n_estimators=100,random_state=0,learning_rate=0.12)
grad.fit(train_lda,y_train)
```

```python
grad.score(test_lda,y_test)
```
