STARTING WITH SEABORN

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data ={"Days":[1,2,3,4,5],
       "NOP":[50,40,60,30,44]}
df=pd.DataFrame(data)

sns.lineplot(data =data, x="Days", y="NOP")
# line graph
plt.show()
```

SEABORN WITH PANDAS

```python
df=pd.read_csv("titanic.csv")
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
df_cleaned = df[(df['Fare'] >= lower_bound) & (df['Fare'] <= upper_bound)]
c=sns.color_palette("viridis")
plt.figure(figsize=[45,10])
sns.lineplot(data=df_cleaned,x="Age",y="Fare", hue="Pclass", palette=c)
plt.show()
```

BAR PLOT

```python
df=sns.load_dataset("tips")
sns.barplot(data=df, x="day", y="tip", estimator="sum", hue="sex", palette="hot", errorbar=("ci",0) )
plt.show()
```

HISTOGRAM

```python
df=pd.read_csv("titanic.csv")
plt.figure(figsize=[20,10])
sns.histplot(data=df,x="Age",hue="Embarked",palette="viridis", kde=True, bins=80)
plt.show()
```

```python
df=sns.load_dataset("tips")
```

SCATTER PLOT

```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="day", palette="viridis",size="size")
plt.show()
```

HEATMAP

```python
gp=df.groupby("day").agg({"tip":"sum"})
sns.heatmap(gp)
plt.show()
```

COUNT PLOT

```python
sns.countplot(data=df, x="size", hue='sex')
```

VIOLIN PLOT

```python
sns.violinplot(data=df, x="size")
plt.plot()
```

PAIR PLOT

```python
sns.pairplot(df, hue="day")
# creates all types of plot 
plt.show()
```

STRIP PLOT

```python
sns.stripplot(data=df, y="total_bill", x="size", hue="smoker", dodge=True, palette="hot")
plt.show()
```

BOX PLOT

```python
sns.boxplot(data=df,y="tip", x="size",palette="spring", hue="sex")
plt.show()
```

CAT PLOT

```python
sns.catplot(data=df, y="tip", x="day", kind="bar",hue="sex", palette="viridis",errorbar=("ci",0) )
# kind gives the type of plot
plt.show()
```

STYLE

```python
df=sns.load_dataset("exercise")
sns.set_style(style= "ticks")
#set styles
sns.barplot(data=df,x="time",y="pulse",errorbar=("ci",0))
plt.show()
```

COLORS

```python
sns.palplot(sns.color_palette())
sns.palplot(sns.color_palette("viridis"))
sns.palplot(sns.color_palette("spring", 3))
# gives the patellle and no. of colors
sns.palplot(sns.color_palette("hot"))
plt.show()
```

MULTIPLE PLOT

```python
df=sns.load_dataset("tips")
```

```python
a= sns.FacetGrid(df, col="day", height=5)
# gives different graphs
a.map(sns.lineplot, "size", "total_bill")
plt.show()
```

RELATIONAL PLOT

```python
sns.relplot(data=df,x="total_bill", y="tip", hue="sex",col="day" ,palette="spring", size="size", kind="scatter")
# default plot is scatter
plt.show()
```

SWARM PLOT

```python
sns.swarmplot(data=df, x="day", y="tip")
#key difference between swarmplot and stripplot is swamplot does not overlap
plt.show()
```

KDE PLOT

```python
#KDE = kernel density estimation 

sns.kdeplot(data=df, x="total_bill", hue="day", multiple="stack",palette="viridis")
```
