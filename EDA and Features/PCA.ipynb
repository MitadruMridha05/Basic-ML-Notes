```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import BaggingRegressor
from sklearn.decomposition import PCA
```

```python
df=pd.read_csv("house_price_regression_dataset.csv")
df.describe()
```

```python
df.info()
```

```python
le=LabelEncoder()
l=le.fit_transform(df["Neighborhood_Quality"])
a=pd.DataFrame(l,columns=["Neighborhood_Quality_N"])
df.drop(columns=["Neighborhood_Quality"],inplace=True)
df["Neighborhood_Quality_N"]=a
```

```python
x=df.drop(columns=["House_Price"])
y=df["House_Price"]
```

```python
sc=StandardScaler()
dfs=sc.fit_transform(x)
```

```python
sns.scatterplot(data=df,x="Lot_Size",y="Year_Built")
plt.show()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
pca=PCA(n_components=3)
pca.fit_transform(x)
pca.explained_variance_
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
param_grid = {
    'fit_intercept': [True, False],
    'positive': [True, False]
}

grid_search = GridSearchCV(lr, param_grid, cv=5, scoring='r2')
grid_search.fit(x_train, y_train)
```

```python
grid_search.best_score_
```

```python
bag_model=BaggingRegressor(lr,n_estimators=200,max_samples=0.8,oob_score=True,random_state=42)
bag_model.fit(x_train,y_train)
bag_model.oob_score_
```
