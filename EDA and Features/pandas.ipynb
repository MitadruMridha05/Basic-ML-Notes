NULL VALUES

```python
import numpy as np

data =pd.read_csv("titanic.csv")
print(data.isnull())# checks for null values
print(data.isnull().sum())# gives no of nulls in each column
print(data.dropna())# deletes null in each column
print(data.replace(np.nan,"oh"))# replaces null value with other value
data["Age"] = data["Age"].replace(np.nan,35) # replaces null in particular column with a particular value
print(data.fillna(method="bfill"))#back_fill ie: fill the upper value at null
print(data["Age"].mean())
print(data.fillna(method="ffill"))#front_fill ie: fill the lower value at null
print(data.fillna("met"))#fills null with given value
```

COLUMN TRANSFORMATION

```python
def ex(v:str) ->str:
    return v[0:3]
data =pd.read_csv("titanic.csv")
data.loc[(data["Age"]>30), "Is_Adult"] = True #creates a new colun in dataframe based on the given conditions
data.loc[(data["Age"]<=30), "Is_Adult"] = False
data["Cabin"]=data["Cabin"].fillna(method='bfill')
data["Passanger_Code"]=data["Name"].map(ex)+data["Cabin"]#concatanates two columns for new column
data['Test']=data["Age"]+data["Fare"]#create new column using operation
print(data.head(6))
```

MERGE JOIN AND CONCATANATE

```python
import pandas as pd
# Create a dictionary with data
data1 = {
    'EmpID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40]
}
# Create a DataFrame
df1 = pd.DataFrame(data1)
# Create a dictionary with data
data2 = {
    'EmpID': [101, 102, 103, 106],
    'Salary': [50000, 60000, 70000, 80000]
}
# Create a DataFrame
df2 = pd.DataFrame(data2)
print(pd.merge(df1,df2,on="EmpID"))#merges common rows in two dataframes
print(pd.merge(left=df1,right=df2,on="EmpID",how="right"))#merges all left dataframe rows and gives NaN in whose values are missing
'''vice versa for right join'''
print(pd.concat([df1,df2]))#concat two same dataframes
```

COMPARE

```python
import pandas as pd
# Create a dictionary with data
data1 = {
    'Fruits': ['Apple', 'Banana', 'Cherry', 'Date'],
    'Price': [100, 30, 150, 200],
    'Quantity': [10, 20, 15, 5]
}
# Create a DataFrame
df1 = pd.DataFrame(data1)
# Display the DataFrame
print(df1)
df2=df1.copy()# copies df1 to df2
df2.loc[0,"Price"]=120 #updates data
df2.loc[1,"Price"]=40
df2.loc[2,"Price"]=170
df2.loc[0,"Quantity"]=12
df2.loc[1,"Quantity"]=23
df2.loc[3,"Quantity"]=4
print(df2)
print(df1.compare(df2,align_axis=0))# compares df1 with df2
print(df1.compare(df2,keep_shape=True))# compares df1 with df2 but shows equal values
```

PIVOTING AND MELTING DATAFRAMES

```python
import pandas as pd
'''Pivoting in pandas is a way to reshape and transform 
DataFrames.Essentially, it allows you to turn unique values 
from one column into separate columns, thus creating a new table
structure. This is particularly useful for data analysis and
visualization.
In more technical terms, the pivot function in pandas produces
a new DataFrame by using unique values from one column as new 
columns, and values from another column as rows'''
data = {
    'EmpID': ['101', '102', '101', '102'],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': ['25', '30', '35', '30'],
    'Salary': [50000, 60000, 70000, 80000]
}
df=pd.DataFrame(data)
print(df)
print(df.pivot(index='EmpID',columns='Name',values=['Age','Salary']))#pivoting
```

MELTING

```python
import pandas as pd
'''Melting a DataFrame in pandas is the process of transforming 
it from a wide format to a long format. This can be particularly 
useful for creating tidy data and preparing it for analysis.'''
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': ['25', '30', '35', '30'],
    'Salary': [50000, 60000, 70000, 80000]
}
df=pd.DataFrame(data)
print(df)
print(pd.melt(df, id_vars=["Name"], value_vars=["Age","Salary"], var_name="AgeSal", value_name="values"))#melting
```
