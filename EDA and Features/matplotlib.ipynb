```python
y=[98,67,88,95,88]
x=['Phsics','Biology','Chemistry','Maths','English']
c=["red","blue","yellow","green","orange"]
plt.bar(x,y,color=c,edgecolor="black")
# plot bar graph where x is x axis and y is y axis with all colours present in c
plt.xlabel("Subjects", fontsize = 20)
# gives label of x-axis
plt.ylabel("Marks Scored",fontsize = 20)
# gives label of y-axis
plt.title("Marks Scored in Different Subjects", fontsize=16)
# gives title of the plot
plt.show() 
#shows the plotted graph
```

VIOLIN PLOT

```python
data= pd.read_csv("titanic.csv")
Q1 = data['Fare'].quantile(0.25)
Q3 = data['Fare'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
data_cleaned = data[(data['Fare'] >= lower_bound) & (data['Fare'] <= upper_bound)]
plt.violinplot(data_cleaned["Fare"], showextrema=True, showmedians=True)
#plots violn_plot
plt.show()
```

STEM PLOT

```python
data= pd.read_csv("titanic.csv")
Q1 = data['Fare'].quantile(0.25)
Q3 = data['Fare'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
data_cleaned = data[(data['Fare'] >= lower_bound) & (data['Fare'] <= upper_bound)]
plt.stem(data_cleaned.head(100)["Fare"],bottom=20 ,linefmt=('r-'), markerfmt='bd', basefmt='g-')
# plots stem with x-axis 
plt.show()
```

STACK PLOT

```python
days=[1,2,3,4,5,6,7]
w_1=[10,20,30,40,50,60,70]
w_2=[70,60,50,40,30,20,10]
w_3=[50,80,10,40,76,34,98]
w_4=[24,5,67,53,47,97,24]
plt.stackplot(days, w_1, w_2, w_3, w_4, colors=["blue","orange","forestgreen","yellowgreen"])
plt.show()
plt.savefig("scene.png")
# saves the plot and facecolor give the background color
```

STEP PLOT

```python
days=[1,2,3,4,5,6,7]
sales=[10,20,30,40,50,60,70]
plt.step(days, sales, where='mid',color="red", marker="o")
```

SUB PLOTS

```python
# Data for plotting
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.sinh(x)

# Create a figure with 4 subplots
fig, axs = plt.subplots(2, 2)

# Plot in each subplot
axs[0, 0].plot(x, y1, 'tab:blue')
axs[0, 0].set_title('Sine Function')
axs[0, 1].plot(x, y2, 'tab:orange')
axs[0, 1].set_title('Cosine Function')
axs[1, 0].plot(x, y3, 'tab:green')
axs[1, 0].set_title('Tangent Function')
axs[1, 1].plot(x, y4, 'tab:red')
axs[1, 1].set_title('Hyperbolic Sine Function')

# Add labels
for ax in axs.flat:
    ax.set(xlabel='x-label', ylabel='y-label')

# Adjust layout to prevent overlapping
plt.tight_layout()
plt.savefig("fig.png")
plt.show()
```
