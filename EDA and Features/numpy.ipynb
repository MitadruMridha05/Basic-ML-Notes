INSPECTION OF DATA

```python
arr= np.array([[10,20,30],[30,50,40]])
print(arr[1,2])
print(np.shape(arr))# dimension
print(np.size(arr))# size 
print(np.ndim(arr)) #no of rows
print(arr.dtype) # datatypes
print(arr.astype(float)) # changes datatype
```

MATHEMATICAL OPERATIONS

```python
a= np.array([10,20,30,80])
b=np.array([30,50,40,10])
print(a+b) # summation
print(np.add(a,b))# add function
'''same for substaraction, multiplication and division'''
print(np.sqrt(b).astype)
```

JOINING AARAY

```python
a= np.array([10,20,30,80])
b=np.array([30,50,40,10])
print(np.concatenate([a,b]))# concatenate
c= np.array([[10,20,30],[30,50,40]])
d=np.array([[10,20,30],[30,50,40]])
print(np.concatenate([c,d],axis=0))# concatenate the verticall 
print(np.concatenate([c,d],axis=1))# concatenate at horizontally
print(np.hstack([c,d]))# concatenate at horizontally
print(np.vstack([c,d]))# concatenate the vertically
```

SPLITTING ARRAY

```python
 
a= np.array([10,20,30,80,90,60])
print(np.array_split(a,2))# splits array according to the parameter number
```

ADD AND REMOVE ARRAY

```python
 
a= np.array([10,20,30,80,90,60])
print(np.append(a,[90,100]))# appends value or array at back
print(np.insert(a,1,60))# 1: array name 2: index 3:value
print(np.insert(a,[1,3],[40])) #insert number at two different index
c= np.array([[10,20,30],[30,50,40]])
print(np.insert(c,[1,2],[40])) # insert value after converting into 1d array
print(np.insert(c,[1,2],[40], axis=0))#insert an array of parameter value
print(np.delete(a,1))# delete value at given index
print(np.delete(c,1))# delete value after converting into 1d array
print(np.delete(c,1, axis=1))# delete value at given index in each row in matrix
```

SORT

```python
 
a= np.array([10,30,20,80,90,60])
print(np.sort(a))# sort function
c= np.array([[10,20,30],[30,50,40]])
print(np.sort(c))# sort all rows in matrix
```

SEARCH

```python
 
a= np.array([10,30,20,80,90,60])
s=np.where(a%20==0)# gives the index of condition in where
print(s)
e=np.sort(a)
print(np.searchsorted(e,3))#Works on sorted array
```

FILTER

```python
 
a= np.array([10,30,20,80])
f=[True,False, True, False]
new=a[f]# prints values of a where index of f = true
print(new)
g=a>25 #gives an bool array which satisfies the conditions
print(g)
print(a[g])
```

AGGREGATE


```python
 
a= np.array([10,30,20,80])
print(np.sum(a))#sum
print(np.min(a))#minimum value
print(np.max(a))#maximum value
print(np.size(a))#size of array
print(np.cumsum(a))#cumulative product
print(np.cumprod(a))#cumulative product
```

STATISTICS

```python
 
import statistics as stat
a= np.array([10,30,20,80])
print(np.mean(a))# mean of array
print(np.median(a))# median of array
print(stat.mode(a))# mode of array
print(np.std(a))# standard deviation of array
print(np.var(a))# variance of array
print(np.corrcoef(a))# correlation coefficient of array
```
