```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
x,y=make_moons(n_samples=300,noise=0.075)
d={"F_1":x[:,0],"F_2":x[:,1],"output":y}
df=pd.DataFrame(d)
df.to_csv("make_moons.csv",index=False)
df.drop(columns=["output"],inplace=True)
```

```python
sns.scatterplot(data=df,x="F_1",y="F_2", c="seagreen")
plt.show()
```

```python
db=DBSCAN(eps=0.25,min_samples=5)
df["output"]=db.fit_predict(df)
```

```python
sns.scatterplot(data=df,x="F_1",y="F_2",hue="output",palette="hot" )
plt.show()
```
