```python
import pandas as pd
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
from mlxtend.frequent_patterns import apriori
```

```python
pd.set_option("display.max_rows",500)
```

```python
df=pd.read_csv("Groceries_dataset.csv")
df.head(10)
```

```python
gp=df.groupby(["itemDescription"]).agg({"itemDescription":"count"})
gp.rename(columns={"itemDescription": "CountOfItemDescription"}, inplace=True)
gp.sort_values(by="CountOfItemDescription",ascending=False)
```

```python
d:dict[int:list]={}
for _,row in df.iterrows():
    key = row[0] 
    value = row[2]
    if key not in d:
        d[key] = []
    d[key].append(value)
market=[]
for i in d.values():
    market.append(i)
del d
market
```

```python
te=TransactionEncoder()
te.fit(market)
```

```python
a=te.transform(market)
data=pd.DataFrame(a,columns=te.columns_)
data
```

```python
apriori(data,min_support=0.05,use_colnames=True).sort_values(by="support",ascending=False)
```
