```python
import pandas as pd
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
```

```python
df=pd.read_csv("Groceries_dataset.csv")
df.head(10)
```

```python
gp=df.groupby(["itemDescription"]).agg({"itemDescription":"count"})
gp.rename(columns={"itemDescription": "CountOfItemDescription"}, inplace=True)
gp.sort_values(by="CountOfItemDescription",ascending=False)
```

```python
d:dict[int:list]={}
for _,row in df.iterrows():
    key=row[0]
    value=row[2]
    if key not in d:
        d[key]=[]
    d[key].append(value)
market=[]
for i in d.values():
    market.append(i)
del d
```

```python
te=TransactionEncoder()
a=te.fit_transform(market)
data=pd.DataFrame(a,columns=te.columns_)
data
```

```python
fpgrowth(data,min_support=0.05,use_colnames=True).sort_values(by="support",ascending=False)
```
