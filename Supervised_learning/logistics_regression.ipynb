```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.read_csv("tips.csv")
df.describe()
```

```python
df["sex"].value_counts()
```

```python
df.describe()
```

```python
sns.scatterplot(data=df,x="total_bill",y="sex")
plt.show()
```

```python
ote=OneHotEncoder(drop="first")
y=ote.fit_transform(df[['sex']]).toarray()
```

```python
x=df[["total_bill"]]
```

OVER SAMPLING

```python
ro=RandomOverSampler()
ro_x,ro_y=ro.fit_resample(x,y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(ro_x,ro_y,test_size=0.2,random_state=302)
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
lr.score(x_test,y_test)
```

```python
sns.scatterplot(data=df,x="total_bill",y="sex")
sns.lineplot(data=df,x="total_bill",y=lr.predict(x),c="orange")
plt.show()
```
