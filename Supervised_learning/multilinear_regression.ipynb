```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
df=pd.read_csv("house_price_regression_dataset.csv")
df.info()
df.describe()
```

```python
df.head(5)
```

```python
x=df[["Square_Footage","Lot_Size"]]
y=df[["House_Price"]]
```

```python
m=MinMaxScaler()
ar=m.fit_transform(x)
x=pd.DataFrame(ar,columns=["Square_Footage","Lot_Size"])
```

```python
ar=m.fit_transform(y)
y=pd.DataFrame(ar,columns=["House_Price"])
```

```python
sns.scatterplot(data=df,x="Square_Footage",y="House_Price")
plt.plot()
```

```python
sns.scatterplot(data=df,x="Lot_Size",y="House_Price")
plt.plot()
df[["Lot_Size","House_Price"]].corr()
```

```python
x.corr()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=253)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```
