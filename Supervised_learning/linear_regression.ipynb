```python
df=pd.read_csv("house_price_regression_dataset.csv")
df.info()
df.describe()
```

```python
df.head(5)
```

```python
x=df[["Square_Footage"]]
y=df[["House_Price"]]
```

```python
sns.scatterplot(data=df,x="Square_Footage",y="House_Price")
plt.plot()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=253)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
lr.predict([34])
```

```python
m=lr.coef_[0]
```

```python
c=lr.intercept_
```

```python
sns.scatterplot(data=df,x="Square_Footage",y="House_Price")
plt.xlabel("Area")
plt.ylabel("Price")
plt.plot(df["Square_Footage"], m*df["Square_Footage"]+c,c="red")
plt.show()
```
